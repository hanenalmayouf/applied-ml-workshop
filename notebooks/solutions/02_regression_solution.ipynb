**السؤال:** هل نستطيع تقدير متوسط قيمة سلة العميل بالريال؟ سنقارن Linear Regression مع توقع ثابت بسيط.

```python
from pathlib import Path
import sys
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src").exists())
sys.path.insert(0, str(ROOT / "src"))
DATA = ROOT / "data" / "raw"
RANDOM_STATE = 42
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split

df = pd.read_parquet(DATA / "manafeth_customers.parquet")
features = ["tenure_months", "orders_per_month", "days_since_last_order", "distinct_categories", "promo_usage_rate"]
clean = df[features + ["avg_basket_sar"]].dropna()
X_train, X_valid, y_train, y_valid = train_test_split(clean[features], clean.avg_basket_sar, test_size=.2, random_state=RANDOM_STATE)
```

## مهمتك

درّب DummyRegressor ثم LinearRegression. احسب MAE للاثنين بالريال. النموذج الجيد يجب أن يخفض الخطأ مقارنة بتوقع الوسيط.

```python
baseline = DummyRegressor(strategy="median").fit(X_train, y_train)
linear = LinearRegression().fit(X_train, y_train)
base_pred = baseline.predict(X_valid)
linear_pred = linear.predict(X_valid)
print("Dummy MAE SAR:", round(mean_absolute_error(y_valid, base_pred), 2))
print("Linear MAE SAR:", round(mean_absolute_error(y_valid, linear_pred), 2))

residuals = y_valid - linear_pred
plt.scatter(linear_pred, residuals, alpha=.2)
plt.axhline(0, color="black")
plt.xlabel("Predicted basket (SAR)")
plt.ylabel("Residual (SAR)")
plt.title("Residuals reveal where the linear model struggles")
plt.show()
```
